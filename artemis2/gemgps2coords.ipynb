```python
import pandas as pd
from pathlib import Path 

def fix_gem_trace_id(gem_num):
    id_dict = {
        '273': '1R.B24LT..HDF',
        '287': '1R.B24RE..HDF',
        '288': '1R.B24..HDF',
        '275': '1R.B24..HDI',
        '296': '1R.B29..HDF',
        '292': '1R.B29..HDI',
        '369': '1R.B12..HDF',
        '290': '1R.B23..HDF',
        '256': '1R.B03..HDF',
        '254': '1R.B03..HDI',
        '246': '1R.B14..HDF',
        '244': '1R.B14..HDI',
        '362': '1R.B01..HDF',
        '251': '1R.B01..HDI',
        '274': '1R.B07.10.HDF',
        '365': '1R.B07..HDF',
        '285': '1R.B20..HDF',
        '361': '1R.B20..HDI',
        '294': '1R.B20.10.HDI',
        '364': '1R.B20W..HDF',
        '366': '1R.B20S..HDF',
    }
    return id_dict[gem_num]

gpsdir = Path('/Volumes/tachyon/Artemis2/10_gemconvert/gps')
gpsfiles = gpsdir.glob('*.txt')
for gpsfile in gpsfiles:
    try:
        df = pd.read_csv(
            gpsfile,
            encoding='utf-8',
            encoding_errors='ignore'
        )
        gem_num = gpsfile.name[0:3]
        gem_id = fix_gem_trace_id(gem_num)
        print(f"{gem_num}, {gem_id}, {df.lat.median()}, {df.lon.median()}")
    except:
        continue
    
```
